## **Alpha-Beta Experiment**

```python
import json
import re
import pandas as pd
from google.colab import files
from IPython.display import display
```

```python
# ============================================================
# 1. UPLOAD ALL JSON SUMMARY FILES
# ============================================================

uploaded = files.upload()

print(f"\nUploaded {len(uploaded)} files.")
```

```python
# ============================================================
# 2. READ ALL FILES
# ============================================================

results = []

for filename, file_content in uploaded.items():

    # --------------------------------------------------------
    # Extract alpha and beta from filename
    #
    # Expected examples:
    # director_evaluation_50_3_summary.json
    # director_evaluation_100_5_summary.json
    # director_evaluation_150_7_summary.json
    # --------------------------------------------------------

    match = re.search(r'(\d+(?:\.\d+)?)_(\d+(?:\.\d+)?)', filename)

    if not match:
        print(f"Skipping {filename}: could not determine alpha/beta")
        continue

    alpha = float(match.group(1))
    beta = float(match.group(2))

    # --------------------------------------------------------
    # Load JSON
    # --------------------------------------------------------

    data = json.loads(file_content.decode("utf-8"))

    results.append({
        "filename": filename,
        "alpha": alpha,
        "beta": beta,
        "data": data
    })


# Sort by beta, then alpha
results = sorted(
    results,
    key=lambda x: (x["beta"], x["alpha"])
)

print(f"Successfully processed {len(results)} experiments.")


# ============================================================
# 3. OVERALL TABLE
# ============================================================

overall_rows = []

for result in results:

    data = result["data"]
    overall = data["overall"]

    overall_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],
        "Overall Precision": overall["precision"],
        "Overall Recall": overall["recall"],
        "Overall F1": overall["f1"]
    })


overall_df = pd.DataFrame(overall_rows)

print("\n" + "=" * 80)
print("TABLE 1 — OVERALL RESULTS")
print("=" * 80)

display(
    overall_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.0f}",
        "Overall Precision": "{:.4f}",
        "Overall Recall": "{:.4f}",
        "Overall F1": "{:.4f}"
    })
)


# ============================================================
# 4. PRECISION TABLE
# ============================================================

precision_rows = []

for result in results:

    categories = result["data"]["categories"]

    precision_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],

        "P Precision": categories["P"]["precision"],
        "I Precision": categories["I"]["precision"],
        "C Precision": categories["C"]["precision"],
        "O Precision": categories["O"]["precision"]
    })


precision_df = pd.DataFrame(precision_rows)

print("\n" + "=" * 80)
print("TABLE 2 — PRECISION BY PICO CATEGORY")
print("=" * 80)

display(
    precision_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.0f}",
        "P Precision": "{:.4f}",
        "I Precision": "{:.4f}",
        "C Precision": "{:.4f}",
        "O Precision": "{:.4f}"
    })
)


# ============================================================
# 5. RECALL TABLE
# ============================================================

recall_rows = []

for result in results:

    categories = result["data"]["categories"]

    recall_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],

        "P Recall": categories["P"]["recall"],
        "I Recall": categories["I"]["recall"],
        "C Recall": categories["C"]["recall"],
        "O Recall": categories["O"]["recall"]
    })


recall_df = pd.DataFrame(recall_rows)

print("\n" + "=" * 80)
print("TABLE 3 — RECALL BY PICO CATEGORY")
print("=" * 80)

display(
    recall_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.0f}",
        "P Recall": "{:.4f}",
        "I Recall": "{:.4f}",
        "C Recall": "{:.4f}",
        "O Recall": "{:.4f}"
    })
)


# ============================================================
# 6. F1 TABLE
# ============================================================

f1_rows = []

for result in results:

    categories = result["data"]["categories"]

    f1_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],

        "P F1": categories["P"]["f1"],
        "I F1": categories["I"]["f1"],
        "C F1": categories["C"]["f1"],
        "O F1": categories["O"]["f1"]
    })


f1_df = pd.DataFrame(f1_rows)

print("\n" + "=" * 80)
print("TABLE 4 — F1 BY PICO CATEGORY")
print("=" * 80)

display(
    f1_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.0f}",
        "P F1": "{:.4f}",
        "I F1": "{:.4f}",
        "C F1": "{:.4f}",
        "O F1": "{:.4f}"
    })
)


# ============================================================
# 7. OPTIONAL: COMBINED DATAFRAME
# ============================================================

print("\n" + "=" * 80)
print("ALL RESULTS LOADED SUCCESSFULLY")
print("=" * 80)

print(f"Number of experiments: {len(results)}")
print(f"Alpha values: {sorted(set(r['alpha'] for r in results))}")
print(f"Beta values : {sorted(set(r['beta'] for r in results))}")
```

## **Steer Value Experiment**

```python
import json
import re
import pandas as pd
from google.colab import files
from IPython.display import display
```

```python
# ============================================================
# 1. UPLOAD ALL JSON SUMMARY FILES
# ============================================================

uploaded = files.upload()

print(f"\nUploaded {len(uploaded)} files.")
```

```python
# ============================================================
# 2. READ ALL FILES
# ============================================================

results = []

for filename, file_content in uploaded.items():

    # --------------------------------------------------------
    # Extract alpha, beta, and value_scale_factor from filename
    #
    # Expected examples:
    # director_evaluation_50_3_0-5_summary.json
    # director_evaluation_100_5_1-0_summary.json
    # director_evaluation_100_5_1-5_summary.json
    # --------------------------------------------------------

    match = re.search(
        r'director_evaluation_(\d+(?:\.\d+)?)_(\d+(?:\.\d+)?)_(\d+)-(\d+)_summary\.json',
        filename
    )

    if not match:
        print(f"Skipping {filename}: could not determine alpha/beta/value_scale")
        continue

    alpha = float(match.group(1))
    beta = float(match.group(2))

    # Convert filename notation:
    # 0-5 -> 0.5
    # 1-0 -> 1.0
    # 1-5 -> 1.5
    value_scale = float(f"{match.group(3)}.{match.group(4)}")

    # --------------------------------------------------------
    # Load JSON
    # --------------------------------------------------------

    data = json.loads(file_content.decode("utf-8"))

    results.append({
        "filename": filename,
        "alpha": alpha,
        "beta": beta,
        "value_scale": value_scale,
        "data": data
    })


# Sort by beta, then alpha, then value scale
results = sorted(
    results,
    key=lambda x: (x["beta"], x["alpha"], x["value_scale"])
)

print(f"Successfully processed {len(results)} experiments.")


# ============================================================
# 3. OVERALL TABLE
# ============================================================

overall_rows = []

for result in results:

    data = result["data"]
    overall = data["overall"]

    overall_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],
        "Value Scale": result["value_scale"],
        "Overall Precision": overall["precision"],
        "Overall Recall": overall["recall"],
        "Overall F1": overall["f1"]
    })


overall_df = pd.DataFrame(overall_rows)

print("\n" + "=" * 80)
print("TABLE 1 — OVERALL RESULTS")
print("=" * 80)

display(
    overall_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.1f}",
        "Value Scale": "{:.1f}",
        "Overall Precision": "{:.4f}",
        "Overall Recall": "{:.4f}",
        "Overall F1": "{:.4f}"
    })
)


# ============================================================
# 4. PRECISION TABLE
# ============================================================

precision_rows = []

for result in results:

    categories = result["data"]["categories"]

    precision_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],
        "Value Scale": result["value_scale"],

        "P Precision": categories["P"]["precision"],
        "I Precision": categories["I"]["precision"],
        "C Precision": categories["C"]["precision"],
        "O Precision": categories["O"]["precision"]
    })


precision_df = pd.DataFrame(precision_rows)

print("\n" + "=" * 80)
print("TABLE 2 — PRECISION BY PICO CATEGORY")
print("=" * 80)

display(
    precision_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.1f}",
        "Value Scale": "{:.1f}",
        "P Precision": "{:.4f}",
        "I Precision": "{:.4f}",
        "C Precision": "{:.4f}",
        "O Precision": "{:.4f}"
    })
)


# ============================================================
# 5. RECALL TABLE
# ============================================================

recall_rows = []

for result in results:

    categories = result["data"]["categories"]

    recall_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],
        "Value Scale": result["value_scale"],

        "P Recall": categories["P"]["recall"],
        "I Recall": categories["I"]["recall"],
        "C Recall": categories["C"]["recall"],
        "O Recall": categories["O"]["recall"]
    })


recall_df = pd.DataFrame(recall_rows)

print("\n" + "=" * 80)
print("TABLE 3 — RECALL BY PICO CATEGORY")
print("=" * 80)

display(
    recall_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.1f}",
        "Value Scale": "{:.1f}",
        "P Recall": "{:.4f}",
        "I Recall": "{:.4f}",
        "C Recall": "{:.4f}",
        "O Recall": "{:.4f}"
    })
)


# ============================================================
# 6. F1 TABLE
# ============================================================

f1_rows = []

for result in results:

    categories = result["data"]["categories"]

    f1_rows.append({
        "Alpha": result["alpha"],
        "Beta": result["beta"],
        "Value Scale": result["value_scale"],

        "P F1": categories["P"]["f1"],
        "I F1": categories["I"]["f1"],
        "C F1": categories["C"]["f1"],
        "O F1": categories["O"]["f1"]
    })


f1_df = pd.DataFrame(f1_rows)

print("\n" + "=" * 80)
print("TABLE 4 — F1 BY PICO CATEGORY")
print("=" * 80)

display(
    f1_df.style
    .format({
        "Alpha": "{:.0f}",
        "Beta": "{:.1f}",
        "Value Scale": "{:.1f}",
        "P F1": "{:.4f}",
        "I F1": "{:.4f}",
        "C F1": "{:.4f}",
        "O F1": "{:.4f}"
    })
)


# ============================================================
# 7. OPTIONAL: COMBINED DATAFRAME
# ============================================================

print("\n" + "=" * 80)
print("ALL RESULTS LOADED SUCCESSFULLY")
print("=" * 80)

print(f"Number of experiments: {len(results)}")
print(f"Alpha values: {sorted(set(r['alpha'] for r in results))}")
print(f"Beta values: {sorted(set(r['beta'] for r in results))}")
print(f"Value scale values: {sorted(set(r['value_scale'] for r in results))}")
```
